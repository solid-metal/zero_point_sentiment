```python
#import library
import numpy as np 
import pandas as pd
```

```python
#load all preprocessed data
amazon_df = pd.read_csv(
    "../processed_data/amazon_processed.csv"
)

student_df = pd.read_csv(
    "../processed_data/student_processed.csv"
)

imdb_df = pd.read_csv(
    "../processed_data/imdb_processed.csv"
)

restaurant_df = pd.read_csv(
    "../processed_data/restaurant_processed.csv"
)
```

```python
#imdb have 50k other have only 1k so imbalance for that make it balance
imdb_df = imdb_df.sample(
    n=5000,
    random_state=42
)
```

```python
#check shape
print(amazon_df.shape)

print(student_df.shape)

print(imdb_df.shape)

print(restaurant_df.shape)
```

```python
#selecting only required column
required_columns = [

    "review",

    "clean_review",

    "sentiment",

    "rating",

    "domain",

    "review_length"
]
```

```python
#keeping only required column
amazon_df = amazon_df[
    required_columns
]

student_df = student_df[
    required_columns
]

imdb_df = imdb_df[
    required_columns
]

restaurant_df = restaurant_df[
    required_columns
]
```

```python
#merging all the datasets
final_df = pd.concat(

    [

        amazon_df,

        student_df,

        imdb_df,

        restaurant_df

    ],

    ignore_index=True
)
```

```python
#checking the data after merging
final_df.head()
```

```python
#check final shape
final_df.shape
```

```python
#checking sentiment
final_df["sentiment"].value_counts()
```

```python
final_df["domain"].value_counts()
```

```python
final_df.drop_duplicates(
    subset=["clean_review"],
    inplace=True
)
```

```python
final_df = final_df[
    final_df["clean_review"] != ""
]
```

```python
# shuffle dataset
final_df = final_df.sample(

    frac=1,

    random_state=42
)
```

```python
final_df.reset_index(
    drop=True,
    inplace=True
)
```

```python
final_df.to_csv(
    "../processed_data/final_merged_dataset.csv",
    index=False
)
```
